A `Document` object in `LangChain` contains information about some data. A Document object has the following two attributes:

- `page_content`: _`str`_: This attribute holds the content of the document\.
- `metadata`: _`dict`_: This attribute contains arbitrary metadata associated with the document. You can use the metadata to track various details, such as the document ID, the file name, and other details.


```python
from langchain_core.documents import Document
```

```python
Doc = Document(
    page_content="""
Python is a high-level, general-purpose programming language known for its simplicity, readability, and versatility. It was created by Guido van Rossum and first released in 1991. Python uses a clear syntax that makes it easier for beginners to understand programming concepts. It supports multiple programming paradigms, including procedural, object-oriented, and functional programming. Today, Python is widely used in web development, automation, data analysis, artificial intelligence, and scientific computing.""",
    metadata={
        "author_id": 20322,
        "author_name": "Hein Thura",
        "release_date": 20 - 11 - 2026,
    },
)

print(Doc.page_content)
```

By using the PDF loader, you can load a PDF file as a Document object.

In this example, you will load the following paper about using LangChain. You can access and read the paper here: Revolutionizing Mental Health Care through LangChain: A Journey with a Large Language Model.


```python
from langchain_community.document_loaders import PyPDFLoader
```

```python
doc = PyPDFLoader(
    file_path="../language_model/pdf/machine_learning_ai.pdf", extraction_mode="plain"
)

pdf = doc.load()
print(pdf[0].page_content)
```

Web Loaders


```python
from langchain_community.document_loaders import WebBaseLoader

loader = WebBaseLoader("https://www.myahtutkhaung.com//")


web_data = loader.load()
print(web_data[0].page_content)
```

Directory Loader


```python
from langchain_community.document_loaders import DirectoryLoader, PyPDFLoader

dir = DirectoryLoader(
    path="../language_model/pdf",
    glob="**/*.pdf",
    loader_cls=PyPDFLoader,
)

docs = dir.load()
print(docs[1].page_content[:1000])
print(docs[2].page_content[:1000])
print(docs[3].page_content[:1000])
```
