```python
from ollama import Client
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
API_KEY = os.getenv("MY_API_KEY")
print(bool(API_KEY))
```

```python
# 1. Select the language model
model_id = "gpt-oss:120b-cloud"

# 2. Configure generation parameters
parameters = {"num_predict": 256, "temperature": 0.2}

# 3. Configure credentials
credentials = {"api_key": API_KEY, "host": "https://ollama.com"}

if not credentials["api_key"]:
    raise ValueError("OLLAMA_API_KEY is missing")

# 4. Initialize the model client
model = Client(
    host=credentials["host"],
    headers={"Authorization": f"Bearer {credentials['api_key']}"},
)

# 5. Generate text
msg = model.generate(
    model=model_id,
    prompt="Complete the following sentence naturally. "
    "Only output the completed sentence, without explanations:\n"
    "In today's sales meeting, we",
    options=parameters,
)

print(msg)
```
