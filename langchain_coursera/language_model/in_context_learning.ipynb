<!-- Zero-shot prompting with ChatOllama -->


```python
import os
from dotenv import load_dotenv
from langchain_ollama import ChatOllama, OllamaLLM
```

```python
load_dotenv()
API_KEY = os.getenv("MY_API_KEY")
```

```python
# create Text_Model

llm = OllamaLLM(
    model="gpt-oss:20b",  # cloud
    base_url="https://ollama.com",
    temperature=0,
    client_kwargs={
        "headers": {
            "Authorization": f"Bearer {API_KEY}",
        }
    },
)
```

### In-context Prompts (examples)


```python
zero_shot_prompt = ["Answer True Or False: The sun rises from East"]
one_shot_prompt = ["""
                Here is an example of translating a sentence from English to French:

                 English: “How is the weather today?”
                French: “Comment est le temps aujourd'hui?”
            
                Now, translate the following sentence from English to French:
            
                 English: “Where is the nearest supermarket?”
                 """]
few_shot_prompt = ["""
                Here is the few examples of classifying emotions in statements:
                Statement: 'I just won my first lottery.'
                Emotion: Joy
                Statement: 'I cannot believe I lost my keys again.'
                Emotion: Frustration
                Statement: 'My best friend is moving to another country.'
                Emotion: Sadness

                Now, classify the emotion in the folliwing statement
                Statement: 'The movie was so scary I have to cover my eyes'
                """]
chain_of_thought = ["""
                A person weighs 70 kg and wants to consume
                1.8 grams of protein per kilogram of body weight.

                Calculate their daily protein requirement.
                Show the formula, calculation, and final answer.
                """]
self_consistency = ["""
                A gym membership costs $40 per month.
                There is a 15 pecent discount for students.
                What is the monthly price after the discount?

                Give me 3 different answers and explanations; and then determine (take out) the most satisfying and consistent reason.
                    """]
```

```python
response1 = llm.invoke(zero_shot_prompt)
print(response1)
```

```python
response2 = llm.invoke(one_shot_prompt)
print(response2)
```

```python
response3 = llm.invoke(few_shot_prompt)
print(response3)
```

```python
response4 = llm.invoke(chain_of_thought)
print(response4)
```

```python
response5 = llm.invoke(self_consistency)
print(response5)
```
