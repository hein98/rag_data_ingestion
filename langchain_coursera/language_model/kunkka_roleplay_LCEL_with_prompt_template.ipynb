```python
import os
from dotenv import load_dotenv
from langchain_ollama import ChatOllama
```

```python
load_dotenv()
API_KEY = os.getenv("MY_API_KEY")
print(bool(API_KEY))
```

### Prompt


```python
from langsmith import Client
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages(
    [
        (
            "system",
            """
        You are Kunkka, the Admiral from Dota 2.

        Personality:
        - Brave, commanding, and confident.
        - Speak like a legendary naval commander.
        - Occasionally use nautical expressions.
        - Have a strong sense of honor and duty.

        Your abilities:
        - Torrent
        - Tidebringer
        - X Marks the Spot
        - Ghostship

        Roleplay rules:
        - Stay in character as Kunkka.
        - Address the user as {player_title}.
        - Respond dramatically but naturally.
        - Keep responses entertaining and concise.

        
        """,
        ),
        ("user", "{user_message}"),
    ]
)
```

```python
# model

llm = ChatOllama(
    model="gpt-oss:20b",
    temperature=0.8,
    base_url="https://ollama.com",
    client_kwargs={"headers": {"Authorization": f"Bearer {API_KEY}"}},
)
```

### StrOutputParser to LCEL chain to get a clean string instead of an AIMessage object.


```python
from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()
```

```python
chain = prompt | llm | parser


result = chain.invoke(
    {
        "player_title": "Hein",
        "user_message": "Hey Kunkka.The enemy team has destroyed all the barracks, now they have Mega Creeps",
    }
)

print(result)
```
