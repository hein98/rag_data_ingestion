```python
import os
from dotenv import load_dotenv
from langchain_ollama import ChatOllama

# API connection
load_dotenv()
API_KEY = os.getenv("MY_API_KEY")
print(bool(API_KEY))

if not API_KEY:
    raise ValueError("MISSING API KEY")
```

```python
# Prompt Formatting with Langchain ChatPromptTemplate

from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages(
    [
        (
            "system",
            "You are an Expert Fitness Coach.Answer using only the provided context.",
        ),
        ("human", "Explain {topic} at {level} level."),
    ]
)
```

```python
# chat_model
from langchain_ollama import ChatOllama

chat_model = ChatOllama(
    model="gpt-oss:120b",
    base_url="https://ollama.com",
    validate_model_on_init=True,
    temperature=0.1,
    top_k=3,
    top_p=0.1,
    # num_predict=256,
    # API_FORMAT
    client_kwargs={"headers": {"Authorization": f"Bearer {API_KEY}"}},
)
```

```python
# parser

from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()


# LCEL
chain = prompt | chat_model | parser
```

```python
# invoke and get result

result1 = chain.invoke({"topic": "Calorie Deficit", "level": "Intermediate"})
print(result1)
```

```python
result2 = chain.invoke({"topic": "Progressive Overload", "level": "Beginner"})
print(result2)
```
