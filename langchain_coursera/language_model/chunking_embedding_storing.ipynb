```python
from langchain_community.document_loaders import PyPDFLoader, WebBaseLoader
```

```python
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/96-FDF8f7coh0ooim7NyEQ/langchain-paper.pdf"

document = PyPDFLoader(url)
docs = document.load()
print(docs[0].page_content)
```

```python
from langchain_text_splitters import (
    CharacterTextSplitter,
    RecursiveCharacterTextSplitter,
)
```

```python
splitter_1 = CharacterTextSplitter(chunk_size=300, chunk_overlap=30, separator="\n\n")
splitter_2 = RecursiveCharacterTextSplitter(
    chunk_size=300, chunk_overlap=30, separators=["\n\n", "\n", " ", ""]
)

chunk_1 = splitter_1.split_documents(docs)
print(chunk_1)
chunk_2 = splitter_2.split_documents(docs)
```

```python
def show_stats(chunks, name):
    lengths = [len(doc.page_content) for doc in chunks]

    print(f"\n{name}")
    print(f"Total chunks: {len(chunks)}")
    print(f"Average size: {sum(lengths) / len(lengths):.2f}")
    print(f"Smallest chunk: {min(lengths)}")
    print(f"Largest chunk: {max(lengths)}")
    print(f"Metadata: {chunks[0].metadata}")
    print(f"Example: {chunks[0].page_content[:150]}")


show_stats(chunk_1, "CharacterTextSplitter")
show_stats(chunk_2, "RecursiveCharacterTextSplitter")
```

```python
from langchain_chroma import Chroma

vector_store = Chroma(
    collection_name="example_collection",
    embedding_function=embeddings,
    persist_directory="./chroma_db",
)

vector_store.add_documents(documents=chunk_2)
```

```python
# Step 3: Check stored documents
print(vector_store._collection.count())

# Step 4: Semantic search
results = vector_store.similarity_search("What is machine learning?", k=3)

# Step 5: Display retrieved chunks
for doc in results:
    print(doc.page_content)
    print("-------------------")
```
