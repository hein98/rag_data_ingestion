```python
from langchain_core.documents import Document
```

```python
doc = Document(
    page_content="This is the main content that I am using to create a RAG system.",
    metadata={
        "source": "my_source",
        "page": 1,
        "author": "Hein",
        "date_created": "2026-10-03",
    },
)

import os

current_directory = os.getcwd()
print("Current Directory:", current_directory)
print("File exists:", os.path.exists("../data/text_files/python_programming.txt"))
```

```python
# TextLoader

from langchain_community.document_loaders import TextLoader

loader = TextLoader("../data/text_files/python_programming.txt", encoding="utf-8")

document = loader.load()
print(document)
```

```python
# Document Structure
# Document Loading _ Parsing

from langchain_community.document_loaders import (
    DirectoryLoader,
    TextLoader,
    PyPDFLoader,
)

print("Import Done")
test_file_directory_loader = DirectoryLoader(
    path="../data/text_files",
    glob="**/*.txt",
    loader_cls=TextLoader,
    loader_kwargs={"encoding": "utf-8"},
    show_progress=True,
)

test_files = test_file_directory_loader.load()
print(test_files)
print(len(test_files))

pdf_file_directory_loader = DirectoryLoader(
    path="../data/pdf_files",
    glob="**/*.pdf",
    loader_cls=PyPDFLoader,
    show_progress=True,
)

pdf_files = pdf_file_directory_loader.load()
print(pdf_files)
print(len(pdf_files))

print("Document Structure Loading and Parsing Done")
```
